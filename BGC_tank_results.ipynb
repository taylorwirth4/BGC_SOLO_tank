## Load data

```python
# prompt: load all csv files from content/drive/my drive/BGC-SOLO-Checkout/data. The name of the csv file corresponds to a serial number. I want each serial number, thus csv file, to have it's own data frame

import pandas as pd
import os

# Mount Google Drive
# from google.colab import drive
# drive.mount('/content/drive')

data_dir = '/content/drive/My Drive/BGC-SOLO-Checkout/data'  # Update with your actual path
dataframes = {}

for filename in os.listdir(data_dir):
    if filename.endswith('.csv'):
        serial_number = filename[:-4]  # Extract serial number (remove .csv extension)
        filepath = os.path.join(data_dir, filename)
        try:
            df = pd.read_csv(filepath)
            df['time'] = pd.to_datetime(df['time'])
            dataframes[serial_number] = df
            print(f"Loaded DataFrame for serial number: {serial_number}")
        except pd.errors.EmptyDataError:
            print(f"Warning: Skipping empty CSV file: {filename}")
        except pd.errors.ParserError:
            print(f"Warning: Could not parse CSV file: {filename}")
        except Exception as e:
            print(f"An error occurred while processing {filename}: {e}")

# Now you can access each dataframe by its serial number
# Example:
# if '12345' in dataframes:
#     df_12345 = dataframes['12345']
#     print(df_12345.head())
```

## Stats code

```python
import numpy as np

def calc_stats(df, variables):
    results = {}
    for var in variables:
        if var in df.columns:
            min_val = np.min(df[var])
            max_val = np.max(df[var])
            avg_val = np.mean(df[var])
            diff = np.max(df[var]) - np.min(df[var])

            # window_size = hours_per_day / (time_interval_minutes / 60)
            # Calculate the time difference between consecutive timestamps
            df['time'] = pd.to_datetime(df['time'])
            time_deltas = df['time'].diff()

            # Drop the first NaT and convert the rest to minutes
            interval_minutes = int(np.floor((time_deltas.dt.total_seconds().dropna() / 60).mean()))
            window_size = int(24 / (interval_minutes / 60))
            window_size = int(interval_minutes * 1)

            smooth = df[var].rolling(window=window_size, center=True, min_periods=1).mean()  # adjust window for your time resolution
            resid = df[var] - smooth
            noise = resid.std()

            # 1. Ensure datetime format
            df['time'] = pd.to_datetime(df['time'])

            # 2. Smooth with a short rolling window (e.g., ±1 hour = 3-hour window)
            smooth = df[var].rolling(window=3, center=True, min_periods=1).mean()

            # 3. Compute residuals
            resid = df[var] - smooth

            # 4. Filter for hours around noon and midnight (±1 hour)
            hr = df['time'].dt.hour
            target_hours = [23, 0, 1, 11, 12, 13]  # midnight and noon ± 1 hour
            subset = resid[hr.isin(target_hours)]

            # 5. Estimate mean absolute residual as proxy for sensor noise
            noise = subset.abs().mean()

            results[var] = {
                'min': min_val,
                'max': max_val,
                'avg': avg_val,
                'diff': diff,
                'noise': noise
            }
        else:
            print(f"Warning: Variable '{var}' not found for float.")
    return results


def format_results(results):
    formatted_data = []
    for var, stats in results.items():
      formatted_data.append([
          var,
          f"{stats['min']:.3g}",
          f"{stats['max']:.3g}",
          f"{stats['avg']:.3g}",
          f"{stats['diff']:.3g}",
          f"{stats['noise']:.3g}"
      ])

    return formatted_data

# Example usage (replace with your desired variables)
variables_to_analyze = ['vac','rh','t', 's', 'p', 'do_umol', 'do_sat', 'do_mL',
                        'phtot', 'phvrs', 'phvk',
                        'chla470', 'chla435', 'bbp700',
                        'chla470_cts', 'chla435_cts', 'bbp700_cts',
                        'irrad380cts', 'irrad443cts', 'irrad490cts', 'irrad555cts',
                        'no3']
stats_frames = {}
for serial_number, df in dataframes.items():
    stats_results = calc_stats(df, variables_to_analyze)
    formatted_stats = format_results(stats_results)

    # Create a pandas DataFrame from the formatted results
    stats_df = pd.DataFrame(formatted_stats, columns=['Variable', 'Avg', 'Min', 'Max', 'Diff', 'Noise'])
    stats_frames[serial_number] = stats_df
# Export to CSV
#stats_df.to_csv('stats_summary.csv', index=False)

# Concatenate all stats_frames into a single DataFrame
all_stats = []
for serial_number, stats_df in stats_frames.items():
    stats_df['Serial Number'] = serial_number  # Add serial number column
    all_stats.append(stats_df)
all_stats_df = pd.concat(all_stats, ignore_index=True)

# Pivot the table
pivoted_stats = all_stats_df.pivot(index='Variable', columns='Serial Number', values=['Avg', 'Min', 'Max', 'Diff', 'Noise'])

```
